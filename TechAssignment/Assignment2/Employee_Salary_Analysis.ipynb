```python
#import libraries
import numpy as np
import pandas as pd
```

```python
#dataset(Salary | Experience (Years) | Performance Score)
salary = np.array([
    [25000, 2, 80],
    [45000, 5, 90],
    [30000, 3, 75],
    [60000, 8, 95],
    [35000, 4, 85]
])

salary
```

```python
#1. Average salary of all employees
average_salary = np.mean(salary[:, 0])
average_salary
```

```python
#2. Highest salary
highest_salary = np.max(salary[:, 0])
highest_salary
```

```python
#3. Lowest salary
lowest_salary = np.min(salary[:, 0])
lowest_salary
```

```python
#4. Average experience
average_experience = np.mean(salary[:, 1])
average_experience
```

```python
#5. Employees with salary greater than ₹40,000
employees = salary[salary[:, 0] > 40000]
employees
```

```python
#6. Employees with performance score greater than 80
employees = salary[salary[:, 2] > 80]
employees
```

```python
#7. Employee with highest performance score
index = np.argmax(salary[:, 2])
employee = salary[index]
employee
```

```python
#8. Standard deviation of salaries
salary_std = np.std(salary[:, 0])
salary_std
```

```python
#9. Classify employees as High Salary or Low Salary
salary_status = np.where(
    salary[:, 0] > 40000,
    "High Salary",
    "Low Salary"
)

salary_status
```

```python
#10. Convert the results into a Pandas DataFrame
df = pd.DataFrame({
    "Salary": salary[:, 0],
    "Experience": salary[:, 1],
    "Performance Score": salary[:, 2],
    "Salary Status": salary_status
})

df
```
